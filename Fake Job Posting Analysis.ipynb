<h4> 10.Handling Missing Values</h4>

```python
df.isnull().sum()
```

```python
#Fill Missing Values (Text Columns)

df['company_profile'] = df['company_profile'].fillna("Not Provided")
df['department']=df['department'].fillna("Not Provided")
df['benefits'] = df['benefits'].fillna("No Benefits")
df['required_education'] = df['required_education'].fillna("Unknown")
df['industry'] = df['industry'].fillna("Unknown")
df['function'] = df['function'].fillna("Unknown")
df['location'] = df['location'].fillna("Unknown")
df['employment_type'] = df['employment_type'].fillna("Unknown")
df['requirements'] = df['requirements'].fillna("No Requirements")
df['description'] = df['description'].fillna("No Description")
```

```python

df['required_experience'] = df['required_experience'].fillna(
    df['required_experience'].mode()[0]
)
```

```python
# Fill the employment type with the mode of the column
df['employment_type'].mode()
mode_value = df['employment_type'].mode()[0]
df['employment_type'] = df['employment_type'].replace("Unknown", mode_value)
```

```python
df['employment_type'].value_counts()
```

<h4>12. Removing unnecessary Columns</h4>

```python
df.drop(columns=['has_questions','function','benefits'])
df = df.drop(columns=['department'])
```

<h4>13. Remove Duplicated Rows</h4>

```python
df.duplicated().sum()
#df.drop_duplicates(inplace=True)
```

<h4>14. Clean Text Data</h4>

```python
df['description'] = df['description'].str.lower()
```

```python
df['description'] = df['description'].str.strip()
df['description'] = df['description'].str.replace(r'[^a-zA-Z0-9 ]', '', regex=True)
```

<h4>15. Handling Salary Range</h4>

```python
df['min_salary'] = df['salary_range'].apply(
    lambda x: str(x).split('-')[0] if '-' in str(x) else None
)
```

```python
df['min_salary'] = pd.to_numeric(df['min_salary'], errors='coerce')
```

```python
df['min_salary'] = df['min_salary'].fillna(df['min_salary'].median())
df['salary_range'] =df['salary_range'].fillna("Not Provided")
```

<h4>16. Cleaning the Location for Map</h4>

```python
df[['country','state','city']] = df['location'].str.split(',', n=2, expand=True)

df['country'] = df['country'].str.strip()
df['state'] = df['state'].str.strip()
df['city'] = df['city'].str.strip()
```

<h4> 17.Replacing the names</h4>

```python
country_map = {
    'GB': 'United Kingdom',
    'NZ': 'New Zealand',
    'CA': 'Canada',
    'SG': 'Singapore',
    'IL': 'Israel',
    'US': 'United States',
    'AU': 'Australia',
    'DE': 'Germany'
}

df['country'] = df['country'].replace(country_map)
```

```python
df[['location','country','state','city']].head(10)
```

```python
df['state'] = df['state'].replace('', 'Unknown')
```

```python
df['city'] = df['city'].replace('', 'Unknown')
```

```python
df['state'] = df['state'].fillna('Unknown')
df['city']  = df['city'].fillna('Unknown')
```

```python
df['desc_length'] = df['description'].astype(str).apply(len)
```

<h4>18. Checking the Null Values</h4>

```python
df.isnull().sum()
```

<b> Full-time jobs: </b>Fraud rates are generally low, with Executive roles at about 4.6% and Mid-Senior level around 4%.<br>
<b> Other employment types:</b>Entry-level positions show about 13.8% fraudulent postings, while Mid-Senior roles have around 5.3%.<br>
<b> Part-time jobs:</b> Mid-Senior level positions show about 19.4% fraud, and jobs with no specific experience requirement show around 12% fraud.<br>
<b> Temporary jobs :</b>Fraudulent postings are almost 0%, likely due to fewer records in the dataset.<br>
<b> Overall Insight:</b>
Fraudulent job postings are more common in contract and part-time jobs, especially at senior or mid-senior experience levels, while full-time jobs show relatively lower fraud rates.

<h3>🔹 3️⃣ MULTIVARIATE ANALYSIS</h3>

### 📊 Pair Plot


```python
# Select numerical columns
numeric_data = df[['telecommuting','has_company_logo','has_questions',
                   'fraudulent','min_salary','desc_length']]

sns.pairplot(numeric_data, hue='fraudulent')

plt.show()
```

#### ✔️ INSIGHT:
The pair plot is used for multivariate analysis. It shows the relationship between multiple numerical variables and also displays how they are distributed.
By coloring based on the fraudulent variable, we can observe how fake and real job postings differ across features like company logo, telecommuting, and description length.

### 📊 Heat Map

```python
num_cols = df.select_dtypes(include='number')

# Calculate correlation matrix
corr_matrix = num_cols.corr()

# Plot heatmap
plt.figure(figsize=(10,7))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)

plt.title("Multivariate Correlation Heatmap")
plt.show()
```

#### ✔️Findings:

•Job description length (desc_length) shows a negative correlation with fraudulent postings, indicating that fake jobs often have shorter descriptions.

•Company logo presence (has_company_logo) has a negative relationship with fraud, suggesting that jobs without a company logo are more likely to be fraudulent.

•Telecommuting jobs (telecommuting) show a slight positive correlation with fraudulent postings, meaning remote jobs may have a higher risk of fraud.

•Jobs with screening questions (has_questions) tend to have a negative correlation with fraud, indicating that legitimate job postings often include screening questions.

•Overall, the heatmap shows that certain job features such as description length, company logo presence, and telecommuting options influence the likelihood of fraudulent job postings.

#### Overview

This section summarizes the key insights obtained from the exploratory data analysis (EDA) of the fake job postings dataset. The objective is to identify patterns and characteristics that distinguish fraudulent job postings from legitimate ones.

#### Key Insights from the Analysis
##### 1. Distribution of Fraudulent vs Real Job Postings

The dataset shows that the majority of job postings are legitimate, while only a small proportion are marked as fraudulent. This indicates a class imbalance where real jobs significantly outnumber fake jobs.

Interpretation:
Although fake jobs are fewer, their presence can still pose serious risks to job seekers.


##### 2. Company Logo Presence

One of the strongest observations from the analysis is that fraudulent job postings are less likely to include a company logo.

Interpretation:
Legitimate companies tend to provide branding elements such as logos, while fake job postings often omit them.

##### 3. Employment Type Patterns

Certain employment types such as contract or unspecified roles show a slightly higher occurrence of fraudulent postings.

Interpretation:
Fraudulent job advertisements may use flexible or vague job types to attract a broader range of applicants.

##### 4. Telecommuting Jobs

Remote or telecommuting jobs appear in both legitimate and fraudulent postings.

Interpretation:
While remote work is not necessarily a sign of fraud, scammers may use remote job offers to attract more applicants.

##### 5. Job Description Characteristics

The analysis of job description length shows that both legitimate and fraudulent jobs may contain short or long descriptions.

Interpretation:
Description length alone is not a strong indicator of job authenticity.

##### 6. Correlation Between Variables

The correlation analysis shows that most variables have weak relationships with the fraudulent label. However, the presence of a company logo shows a slightly stronger negative correlation with fraudulent postings.

Interpretation:
This suggests that features such as company logo availability may help in identifying suspicious job postings.

#### Overall Findings

From the exploratory data analysis, several patterns were identified that can help distinguish fraudulent job postings from legitimate ones. Features such as the presence of company logos, employment types, and certain job posting characteristics provide useful signals for identifying potential fraud.
